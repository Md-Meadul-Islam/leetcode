1722. Minimize Hamming Distance After Swap Operations


```python
from collections import defaultdict, Counter
from typing import List
class Solution:
    def minimumHammingDistance(self, source: List[int], target: List[int], allowedSwaps: List[List[int]]) -> int:
        n = len(source)
        parent = list(range(n))
        print(parent)
        
        def find(x):
            if parent[x] != x:
                parent[x] = find(parent[x])
            return parent[x]
        
        def union(x, y):
            px, py = find(x), find(y)
            if px != py:
                parent[px] = py
        
        # Step 1: union
        for a, b in allowedSwaps:
            union(a, b)
        print(parent)
        
        # Step 2: group indices
        groups = defaultdict(list)
        for i in range(n):
            groups[find(i)].append(i)
        print(groups)
        # Step 3: count mismatches
        res = 0
        
        for indices in groups.values():
            count = Counter()
            
            for i in indices:
                count[source[i]] += 1
            
            for i in indices:
                if count[target[i]] > 0:
                    count[target[i]] -= 1
                else:
                    res += 1
        
        return res

    def minimumHammingDistance2(self, source, target, allowedSwaps):
        n = len(source)
        
        parent = list(range(n))
        
        def find(x):
            while parent[x] != x:
                x = parent[x]
            return x
        
        def union(x, y):
            px = find(x)
            py = find(y)
            if px != py:
                parent[px] = py
        
        # connect indices
        for a, b in allowedSwaps:
            union(a, b)
        
        # group indices
        groups = defaultdict(list)
        for i in range(n):
            groups[find(i)].append(i)
        
        # count mismatches
        res = 0
        
        for indices in groups.values():
            src_vals = []
            tgt_vals = []
            
            for i in indices:
                src_vals.append(source[i])
                tgt_vals.append(target[i])
            
            for val in src_vals:
                if val in tgt_vals:
                    tgt_vals.remove(val)
                else:
                    res += 1
        
        return res
```

```python
obj = Solution()
source = [1,2,3,4]
target = [2,1,4,5]
allowedSwaps = [[0,1],[2,3]]
obj.minimumHammingDistance(source, target, allowedSwaps)
```
