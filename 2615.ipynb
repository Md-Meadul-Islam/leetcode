2615. Sum of Distances


```python
from typing import List
from collections import defaultdict
class Solution:
    def distance_brute_force(self, nums: List[int]) -> List[int]:

        hash_map = defaultdict(list)
        for i, num in enumerate(nums):
            hash_map[num].append(i)
        res = [0]*len(nums)
        for i, num in enumerate(nums):
            arr = hash_map.get(num)
            if len(arr)>1:
                curr_sum = 0
                for j in arr:
                    curr_sum += abs(i-j)
                res[i] = curr_sum
        return res
    
    def distance_prefix_sum(self, nums: List[int]) -> List[int]:

        hash_map = defaultdict(list)
        for i, num in enumerate(nums):
            hash_map[num].append(i)
        res = [0]*len(nums)

        for num, arr in hash_map.items():
            n = len(arr)
            prefix = [0]*n
            prefix[0] = arr[0]
            for i in range(1, n):
                prefix[i] = prefix[i-1] + arr[i]
            for i in range(n):
                idx = arr[i]
                left = idx * i - (prefix[i-1] if i > 0 else 0)
                right = (prefix[n-1]-prefix[i]) - idx*(n-i-1)
                res[idx] = left + right
        return res
    
    def distance_single_pass(self, nums: List[int]) -> List[int]:

        hash_map = defaultdict(list)
        for i, num in enumerate(nums):
            hash_map[num].append(i)
        res = [0]*len(nums)

        for arr in hash_map.values():
            left_sum = 0
            left_count = 0
            right_sum = sum(arr)
            right_count = len(arr)

            for idx in arr:
               # remove current from right side
               right_sum -= idx
               right_count -= 1

               left = idx * left_count - left_sum
               right = right_sum - idx * right_count
               res[idx] = left + right

               # add to left side
               left_sum += idx
               left_count += 1
        return res
    
    def distance(self, nums: List[int]) -> List[int]:
        d=defaultdict(list)
        for i,num in enumerate(nums):
            d[num].append(i)
        a=[0]*len(nums)
        for values in d.values():
            if len(values)>1:
                c=len(values)
                idx=values[0]
                a[idx]=sum(values)-c*idx
                i,j=0,c-2
                for n in values[1:]:
                    a[n]=a[idx]+(i-j)*(n-idx)
                    i+=1
                    j-=1
                    idx=n
        return a
```

```python
obj = Solution()
nums = [1,3,1,1,2]
obj.distance_single_pass(nums)
```
