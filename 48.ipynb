48. Rotate Image


```python
from typing import List
class Solution:
    def rotate2(self, matrix: List[List[int]]) -> None:
        n = len(matrix)
        new_mat = [[None for _ in range(n)] for _ in range(n)]
        for i in range(n):
            for j in range(n):
                val = matrix[i][j]
                new_mat[j][n-i-1] = val
        print(new_mat)
    
    def rotate(self, matrix: List[List[int]]) -> None:
        n = len(matrix)
        def swap(i, j):
            temp = matrix[i][j]
            matrix[i][j] = matrix[j][i]
            matrix[j][i] = temp
            
        for i in range(n):
            for j in range(i):
                swap(i, j)

        for i in range(n):
            matrix[i] = matrix[i][::-1]
```

```python
obj = Solution()
mat = [[1,2,3],[4,5,6],[7,8,9]]
obj.rotate(mat)
```
