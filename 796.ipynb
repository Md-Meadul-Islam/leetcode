796. Rotate String


```python
class Solution:
    def rotateString(self, s: str, goal: str) -> bool:
        if len(s)!= len(goal):
            return False
        
        perimeter=s+s
        if goal in perimeter:
            return True 
        return False
    def rotateString_brute_force(self, s: str, goal: str) -> bool:
        n = len(s)
        i = 0
        while i < n:
            char = s[0:i+1]
            new_str = s[i+1:] + char
            if new_str == goal:
                return True
            i += 1
        return False
```

```python
obj = Solution()
obj.rotateString(s = "abcde", goal = "cdeab")
```
