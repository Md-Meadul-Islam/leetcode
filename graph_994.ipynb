994. Rotting Oranges


```python
from typing import List
from collections import deque
class Solution:
    def orangesRotting(self, grid: List[List[int]]) -> int:
        queue = deque()
        fresh = 0
        col = len(grid)
        row = len(grid[0])

        for i in range(col):
            for j in range(row):
                if grid[i][j] == 2:
                    queue.append((i, j))
                elif grid[i][j] == 1:
                    fresh += 1
        minutes = 0
        directions = [(1,0), (-1,0), (0,1), (0,-1)]

        while queue and fresh > 0:
            for _ in range(len(queue)):
                i, j = queue.popleft()
                for dx, dy in directions:
                    ni, nj = i + dx, j + dy
                    if 0 <= ni < col and 0 <= nj < row and grid[ni][nj] == 1:
                        grid[ni][nj] = 2 #rot
                        fresh -= 1
                        queue.append((ni, nj))
            minutes += 1
        return minutes if fresh == 0 else -1




        
```

```python
grid = [[2,1,1],[1,1,0],[0,1,1]]
obj = Solution()
obj.orangesRotting(grid)
```
