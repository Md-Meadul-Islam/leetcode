3761. Minimum Absolute Distance Between Mirror Pairs


```python
from typing import List
from collections import defaultdict
class Solution:
    def reverse(self, num:int):
        new_num = 0
        while num > 0:
            new_num = (new_num*10) +  (num % 10) 
            num = num // 10
        return new_num


    def minMirrorPairDistance(self, nums: List[int]) -> int:
        reverseMap = defaultdict()
        minDistance = float("inf")
        for i, num in enumerate(nums):
            reverseNum = int(str(num)[::-1])
            if num in reverseMap:
                minDistance = min(minDistance, abs(reverseMap.get(num)- i))
            reverseMap[reverseNum] = i
        print(reverseMap)
        return -1 if minDistance == float('inf') else minDistance
            

```
