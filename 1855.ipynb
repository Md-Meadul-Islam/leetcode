1855. Maximum Distance Between a Pair of Values


```python
from typing import List
class Solution:
    def maxDistance(self, nums1: List[int], nums2: List[int]) -> int:
        n = len(nums1)
        m = len(nums2)
        i=0
        j = 0
        max_dist = 0
        while i<n and j<m:
            if nums1[i] > nums2[j]:
                i += 1
            else:
                max_dist = max(max_dist, (j - i))
                j += 1
        return max_dist
        
```

```python
obj = Solution()
nums1 = [55,30,5,4,2]
nums2 = [100,20,10,10,5]
obj.maxDistance(nums1, nums2)
```
