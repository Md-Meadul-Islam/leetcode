3488. Closest Equal Element Queries


```python
from typing import List
from collections import defaultdict
from bisect import bisect_left
class Solution:
    def solveQueries_brute_force(self, nums: List[int], queries: List[int]) -> List[int]:
        pos = defaultdict(list)
        num_len = len(nums)
        for i, n in enumerate(nums):
            pos[n].append(i)

        result = []
        for q in queries:
            query = nums[q]
            indices = pos.get(query)
            min_dist = float('inf')

            if len(indices) == 1:
                result.append(-1)
                continue

            for j in indices:
                if j == q:
                    continue

                if nums[j] == query:
                    d = abs(q-j)
                    d = min(d, num_len-d)
                    min_dist = min(min_dist, d)

            if min_dist == float('inf'):
                result.append(-1)
            else:
                result.append(min_dist)
        return result

    def solveQueries_binary_search(self, nums: List[int], queries: List[int]) -> List[int]:
        pos = defaultdict(list)
        num_len = len(nums)
        for i, n in enumerate(nums):
            pos[n].append(i)

        result = []
        for q in queries:
            query = nums[q]
            indices = pos.get(query)

            if len(indices) == 1:
                result.append(-1)
                continue
            
            idx = bisect_left(indices, q)
            left = indices[idx -1] if idx > 0 else indices[-1]
            right = indices[idx+1] if idx < len(indices)-1 else indices[0]

            d1 = abs(left- q)
            d2 = abs(right- q)

            d1 = min(d1, num_len - d1)
            d2 = min(d2, num_len - d2)
            result.append(min(d1,d2))

        return result

    def solveQueries(self, nums: List[int], queries: List[int]) -> List[int]:
        N = len(nums)
        ans = []
        first = {}
        record = {}

        for i in range(N):
            if nums[i] not in record:
                ans.append(float("inf"))
                record[nums[i]] = i
                first[nums[i]] = i

            else:
                dist = i - record[nums[i]]
                ans.append(dist)
                ans[record[nums[i]]] = min(ans[record[nums[i]]], dist)
                record[nums[i]] = i
        print(ans)
        print(first)
        print(record)

            
        # Explicit circular case.
        for num in first:
            if first[num] != record[num]:
                dist = N - record[num] + first[num]
                ans[record[num]] = min(ans[record[num]], dist)
                ans[first[num]] = min(ans[first[num]], dist)

        print("ans", ans)

        res = []
        for index in queries:
            if ans[index] == float("inf"):
                ans[index] = -1
            res.append(ans[index])

        return res
                
                
```

```python
obj = Solution()
nums = [1,3,1,4,1,3,2]
queries = [0,3,5]
obj.solveQueries_brute_force(nums, queries)
obj.solveQueries(nums, queries)
```
