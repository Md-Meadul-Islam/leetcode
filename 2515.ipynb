```python
from typing import List
class Solution:
    def closestTarget(self, words: List[str], target: str, startIndex: int) -> int:
        min_steps = 100
        words_len = len(words)
        found = False
        for i, word in enumerate(words):
            if word == target:
                found = True
                diff = abs(i- startIndex)
                min_steps = min(min_steps, diff, words_len - diff)
        return min_steps if found else -1
        
    def closestTarget2(self, words: list[str], target: str, startIndex: int) -> int:
        n = len(words)
        return min((min(abs(i - startIndex), n - abs(i - startIndex)) for i, w in enumerate(words) if w == target), default=-1)
```

```python
words = ["a","b","leetcode"]
target = "leetcode"
startIndex = 0
obj = Solution()
obj.closestTarget(words, target, startIndex)
```
