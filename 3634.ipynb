3634. Minimum Removals to Balance Array

```python
from typing import List
class Solution:
    def minRemoval(self, nums: List[int], k: int) -> int:

        nums.sort()
        left = 0
        right = 0
        n= len(nums)
        max_len = 0
        while right < n:
            while nums[left] * k < nums[right]:
                left += 1
            max_len = max(max_len, right - left + 1)
            right += 1
        return n - max_len

        
```
