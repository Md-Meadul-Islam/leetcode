60. Permutation Sequence


```python
class Solution:
    def getPermutation(self, n: int, k: int) -> str:
        result = []
        kth= ""
        count = 0
        def backtrack(path, used):
            nonlocal count
            if count == k:
                kth = path[:]
                return
            count += 1
            for i in range(n):
                if used[i+1]:
                    continue

                path = path + str(i+1)
                used[i+1] =True
                backtrack(path, used)
                path = path[:-1]
                used[i+1] = False

        backtrack("", [False]*(n+1))
        return kth
        
```

```python
obj = Solution()
obj.getPermutation(3, 3)
```
