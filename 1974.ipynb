1974. Minimum Time to Type Word Using Special Typewriter


```python
class Solution:
    def minTimeToType(self, word: str) -> int:
        alphas = {"a":1, "b":2, "c":3, "d":4, "e":5,
                  "f":6,"g":7, "h":8, "i":9, "j":10, "k":11,
                  "l":12,"m":13, "n":14,"o":15,"p":16, "q":17,
                  "r":18, "s":19, "t":20,"u":21,"v":22,"w":23, "x":24, "y":25, "z":26}
        
        seconds = 0
        char = "a"
        for i in range(len(word)):
            next_char = word[i]
            diff = abs(alphas[char] - alphas[next_char])
            steps = min(diff, 26 - diff)
            seconds += steps + 1
            char = next_char
        return seconds
    
    def minTimeToType2(self, word: str) -> int:
        curr = 'a'
        seconds = 0

        for ch in word:
            diff = abs(ord(ch) - ord(curr))
            steps = min(diff, 26 - diff)
            seconds += steps + 1

            curr = ch

        return seconds

```
