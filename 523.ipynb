523. Continuous Subarray Sum

```python
from typing import List

class Solution:
    def checkSubarraySum_set(self, nums: List[int], k: int) -> bool:
        
        # just need to know if there is exact SUM! needs to be of size
        # '2' just don't update the prefixMap until the NEXT iteration.
        seen = set()
        prefixRem = lastPrefixRem = 0
        for v in nums:
            prefixRem = (prefixRem + v) % k
            if prefixRem in seen:
                print(seen)
                return True
            # needs to be SIZE two. SO UPDATE the mp with i-1 prefixRem!           
            seen.add(lastPrefixRem)
            lastPrefixRem = prefixRem
        print(seen)
        return False

    def checkSubarraySum(self, nums:List[int], k:int)->bool:
        # Edge case: k == 0
        if k == 0:
            # We need at least two consecutive zeros
            for i in range(len(nums) - 1):
                if nums[i] == 0 and nums[i + 1] == 0:
                    return True
            return False
        prefix_map = {0: -1}  # remainder -> earliest index
        curr_sum = 0

        for i, num in enumerate(nums):
            curr_sum += num
            rem = curr_sum % k

            if rem in prefix_map:
                # Check subarray length >= 2
                if i - prefix_map[rem] >= 2:
                    return True
            else:
                # Store only the first occurrence
                prefix_map[rem] = i

        return False
```
