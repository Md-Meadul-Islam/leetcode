2452. Words Within Two Edits of Dictionary


```python
from typing import List
import collections
class Solution:
    def twoEditWords(self, queries: List[str], dictionary: List[str]) -> List[str]:
        def matcher(query, dic):
            not_matched_indexes = 0
            for i in range(len(query)):
                if query[i] != dic[i]:
                    not_matched_indexes += 1
                    if not_matched_indexes > 2:
                        return False
            return True

        ans = []
        for query in queries:
            for dic in dictionary:
                if len(query) == len(dic):
                    if matcher(query, dic):
                        ans.append(query)
                        break
        return ans
                
```

```python
queries = ["tsl","sri","yyy","rbc","dda","qus","hyb","ilu","ahd"]
dictionary = ["uyj","bug","dba","xbe","blu","wuo","tsf","tga"]
obj = Solution()
obj.twoEditWords(queries, dictionary)
```
