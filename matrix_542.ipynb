```python
from typing import List
from collections import deque
import math

class Solution:
    def updateMatrix_Fastest(self, mat: List[List[int]]) -> List[List[int]]:
        m, n = len(mat), len(mat[0])

        for r in range(m):
            for c in range(n):
                if mat[r][c] > 0:
                    top = mat[r - 1][c] if r > 0 else math.inf
                    left = mat[r][c - 1] if c > 0 else math.inf
                    mat[r][c] = min(top, left) + 1

        for r in range(m - 1, -1, -1):
            for c in range(n - 1, -1, -1):
                if mat[r][c] > 0:
                    bottom = mat[r + 1][c] if r < m - 1 else math.inf
                    right = mat[r][c + 1] if c < n - 1 else math.inf
                    mat[r][c] = min(mat[r][c], bottom + 1, right + 1)

        return mat
    def updateMatrix_DP(self, mat: List[List[int]]) -> List[List[int]]:
        col = len(mat)
        row = len(mat[0])
        INT_MAX = float('inf')
        dp = [[INT_MAX for _ in range(row)] for _ in range(col)]
        for i in range(col):
            for j in range(row):
                if mat[i][j] == 0:
                    dp[i][j] = 0
        for i in range(col):
            for j in range(row):
                if mat[i][j] != 0:
                    if i > 0:
                        dp[i][j] = min(dp[i][j], dp[i-1][j] + 1)#curr, left
                    if j > 0:
                        dp[i][j] = min(dp[i][j], dp[i][j-1] + 1)#curr, top
        for i in range(col-1, -1, -1):
            for j in range(row-1, -1, -1):
                if mat[i][j] != 0:
                    if i < col-1:
                        dp[i][j] = min(dp[i][j], dp[i+1][j] + 1)#curr, right
                    if j < row-1:
                        dp[i][j] = min(dp[i][j], dp[i][j+1] + 1)#curr, bottom
        return dp
        


    def updateMatrix_BFS(self, mat):
        rows, cols = len(mat), len(mat[0])
        queue = deque()

        # Step 1: initialize
        for i in range(rows):
            for j in range(cols):
                if mat[i][j] == 0:
                    queue.append((i, j))
                else:
                    mat[i][j] = -1

        # Step 2: BFS
        directions = [(1,0), (-1,0), (0,1), (0,-1)]

        while queue:
            i, j = queue.popleft()

            for dx, dy in directions:
                ni, nj = i + dx, j + dy

                if 0 <= ni < rows and 0 <= nj < cols and mat[ni][nj] == -1:
                    mat[ni][nj] = mat[i][j] + 1
                    queue.append((ni, nj))

        return mat

```
