37. Sudoku Solver


```python
from typing import List
class Solution:
    def isValid(self, row, col, val, board):
        # check row
        for r in range(9):
            if board[row][r] == val:
                return False
        #check col
        for c in range(9):
            if board[c][col] == val:
                return False
        #chek box 3*3
        start_r = (row//3) * 3
        start_c = (col//3) * 3
        for i in range(3):
            for j in range(3):
                if board[start_r + i][start_c+ j] == val:
                    return False
        return True

    def solveSudoku_brute_force(self, board: List[List[str]]) -> None:
        def backtrack():
            for r in range(9):
                for c in range(9):
                    if board[r][c] == ".":
                        for num in map(str, range(1,10)):#generate 1 to 9 in str
                            if self.isValid(r, c, num, board):#check the number is valid
                                board[r][c] = num
                                if backtrack(board):
                                    return True
                                board[r][c] = "."#undo
                        return False#no valid number
            return True

        backtrack(board)


    def solveSudoku(self, board:List[List[str]])->None:
        def backtrack(row, col):
            if row == 9:
                return True
            nextRow = row
            nextCol = col + 1
            if nextCol == 9:
                nextRow = row + 1
                nextCol = 0
            if board[row][col] != ".":
                return backtrack(nextRow, nextCol)
            
            for num in map(str, range(1, 10)):
                if self.isValid(row, col, num, board):
                    board[row][col] = num
                    if backtrack(nextRow, nextCol):
                        return True
                    board[row][col] = "."
            return False

    
        backtrack(0, 0)
        print(board)



```
