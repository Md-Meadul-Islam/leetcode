```python
from typing import List
class Solution:
    def isValid(self, row, col, val, board):
        # check row
        for r in range(9):
            if board[row][r] == val:
                return False
        #check col
        for c in range(9):
            if board[c][col] == val:
                return False
        #chek box 3*3
        start_r = (row//3) * 3
        start_c = (col//3) * 3
        for i in range(3):
            for j in range(3):
                if board[start_r + i][start_c+ j] == val:
                    return False
        return True
    def isValid2(self, row, col, val, board):
        for i in range(9):
            if board[row][i] == val:
                return False
            if board[i][col] == val:
                return False
            if board[3 * (row//3) + i //3][3*(col//3)+i%3] == val:
                return False
        return True

    def solveSudoku_brute_force(self, board: List[List[str]]) -> None:
        def backtrack():
            for r in range(9):
                for c in range(9):
                    if board[r][c] == ".":
                        for num in map(str, range(1,10)):#generate 1 to 9 in str
                            if self.isValid(r, c, num, board):#check the number is valid
                                board[r][c] = num
                                if backtrack(board):
                                    return True
                                board[r][c] = "."#undo
                        return False#no valid number
            return True

        backtrack(board)


    def solveSudok_row_wise(self, board:List[List[str]])->None:
        def backtrack(row, col):
            if row == 9:
                return True
            nextRow = row
            nextCol = col + 1
            if nextCol == 9:
                nextRow = row + 1
                nextCol = 0
            if board[row][col] != ".":
                return backtrack(nextRow, nextCol)
            
            for num in map(str, range(1, 10)):
                if self.isValid2(row, col, num, board):
                    board[row][col] = num
                    if backtrack(nextRow, nextCol):
                        return True
                    board[row][col] = "."
            return False

    
        backtrack(0, 0)
    def solveSudoku(self, board):
        rows = [set() for _ in range(9)]
        cols = [set() for _ in range(9)]
        boxes = [set() for _ in range(9)]

        for r in range(9):
            for c in range(9):
                if board[r][c] != ".":
                    val = board[r][c]
                    rows[r].add(val)
                    cols[c].add(val)
                    boxes[(r // 3) * 3 + (c // 3)].add(val)
        print(rows)
        print(cols)
        print(boxes)

        def backtrack(r, c):
            if r == 9:
                return True

            next_r = r + (c + 1) // 9
            next_c = (c + 1) % 9

            if board[r][c] != ".":
                return backtrack(next_r, next_c)

            for num in map(str, range(1, 10)):
                box_idx = (r // 3) * 3 + (c // 3)

                if num in rows[r] or num in cols[c] or num in boxes[box_idx]:
                    continue

                # try new char
                board[r][c] = num
                rows[r].add(num)
                cols[c].add(num)
                boxes[box_idx].add(num)

                if backtrack(next_r, next_c):
                    return True

                # remove
                board[r][c] = "."
                rows[r].remove(num)
                cols[c].remove(num)
                boxes[box_idx].remove(num)

            return False

        # backtrack(0, 0)



```

```python
def solveSudoku_bit_mask(board):
    rows = [0] * 9
    cols = [0] * 9
    boxes = [0] * 9

    # Initialize masks
    for r in range(9):
        for c in range(9):
            if board[r][c] != ".":
                num = int(board[r][c])
                mask = 1 << num
                rows[r] |= mask
                cols[c] |= mask
                boxes[(r // 3) * 3 + (c // 3)] |= mask
    

    def backtrack(r, c):
        if r == 9:
            return True

        next_r = r + (c + 1) // 9
        next_c = (c + 1) % 9

        if board[r][c] != ".":
            return backtrack(next_r, next_c)

        box_idx = (r // 3) * 3 + (c // 3)

        for num in range(1, 10):
            mask = 1 << num

            if (rows[r] & mask) or (cols[c] & mask) or (boxes[box_idx] & mask):
                continue

            # place
            board[r][c] = str(num)
            rows[r] |= mask
            cols[c] |= mask
            boxes[box_idx] |= mask

            if backtrack(next_r, next_c):
                return True

            # remove (backtrack)
            board[r][c] = "."
            rows[r] ^= mask
            cols[c] ^= mask
            boxes[box_idx] ^= mask

        return False

    backtrack(0, 0)
    
```
