7. Reverse Integer

```python
class Solution:
    def reverse_overflow_check_after(self, x: int) -> int:
        reverse_num = 0
        is_negative = False
        if x < 0:
            is_negative = True
            x = -x
        
        while x:
            reverse_num = (reverse_num * 10) + (x % 10)
            x = x // 10
        if is_negative:
            reverse_num = -reverse_num

        if reverse_num < -2147483648 or reverse_num > 2147483647:#2**31=2147483648
            return 0
        return reverse_num
    def reverse_str(self, x: int) -> int:
        x = str(x)

        is_negative = False
        if x[0] == "-":
            x = x[1:]
            is_negative = True

        x = int(x[::-1])

        if is_negative == True:
            x = 0-x

        if x > 2147483647 or x < -2147483648:
            return 0
        else:
            return x
    def reverse_overflow_check_before(self, x: int) -> int:
        sign = -1 if x < 0 else 1
        x = abs(x)
        rev = 0

        while x:
            digit = x % 10
            x //= 10

            if rev > 2147483647 // 10:
                return 0

            rev = rev * 10 + digit

        return sign * rev
```
