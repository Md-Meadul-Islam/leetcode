287. Find the Duplicate Number


```python
from typing import List
class Solution:
    def findDuplicate(self, nums: List[int]) -> int:
    # Phase 1: detect cycle
        slow = nums[0]
        fast = nums[0]

        while True:
            slow = nums[slow]
            fast = nums[nums[fast]]
            if slow == fast:
                break

        # Phase 2: find cycle entry (duplicate)
        slow = nums[0]
        while slow != fast:
            slow = nums[slow]
            fast = nums[fast]

        return slow
    
    def findDuplicate_binary_search(nums):
        low, high = 1, len(nums) - 1

        while low < high:
            mid = (low + high) // 2

            count = 0
            for num in nums:
                if num <= mid:
                    count += 1

            if count > mid:
                high = mid      # duplicate in left
            else:
                low = mid + 1   # duplicate in right

        return low
    def findDuplicate3(self, nums: List[int]) -> int:
        visit=[-1]*len(nums)
        for num in nums:
            if visit[num]==-1:
                visit[num]=1
            else:
                return num
    def findDuplicate_bit_mask(nums):
        n = len(nums) - 1
        duplicate = 0

        for bit in range(32):  # check all bits
            mask = 1 << bit

            nums_count = 0
            expected_count = 0

            for num in nums:
                if num & mask:
                    nums_count += 1

            for i in range(1, n + 1):
                if i & mask:
                    expected_count += 1

            if nums_count > expected_count:
                duplicate |= mask

        return duplicate
```
