<h2>22. Generate Parentheses</h2>
<h4> Catalan Recurrence Formula <h4>
dp[k] = for i in range(k):<br>
"(" + dp[i] + ")" + dp[k-1-i]


```python
from typing import List
class Solution:
    def generateParenthesis_backtracking(self, n: int) -> List[str]:
        fin_ans = []
        def backtrack(open, close, path):
            if len(path) == 2*n:
                fin_ans.append(path)
                return
            if open < n:
                backtrack(open +1, close, path + "(")
            if close < open:
                backtrack(open, close+1, path + ")")
        backtrack(0, 0, "")
        return fin_ans
```

```python
obj = Solution()
obj.generateParenthesis_backtracking(3)
```

```python
def generateParenthesis_dp(n:int)->List[str]:
    dp = [[] for _ in range(n+1)]
    dp[0] = [""]
    for k in range(1, n+1):
        current = []
        for i in range(k):
            for inside in dp[i]:
                for outside in dp[k-1-i]:
                    current.append("(" + inside+ ")" + outside)
        dp[k] = current
    print(dp)
generateParenthesis_dp(3)
```
