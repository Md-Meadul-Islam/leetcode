239. Sliding Window Maximum


```python
from typing import List
from collections import deque
import heapq#min heap priority queue

class Solution:
    def maxSlidingWindow_brute_force(self, nums: List[int], k: int) -> List[int]:
        left = 0
        right = left + k
        total = len(nums) + 1
        res = []
        while right < total:
            res.append(max(nums[left:right]))
            left += 1
            right += 1
        return res
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        q = deque()  # stores indices
        res = []

        for i in range(len(nums)):
            # 1. Remove out-of-bound indices
            if q and q[0] <= i - k:
                q.popleft()

            # 2. Maintain decreasing order
            while q and nums[q[-1]] < nums[i]:
                q.pop()

            # 3. Add current index
            q.append(i)

            # 4. Add to result once window is full
            if i >= k - 1:
                res.append(nums[q[0]])
        return res
    def maxSlidingWindow_heap(self, nums: List[int], k: int) -> List[int]:
        heap = []  # max heap (store negative values)
        res = []

        for i in range(len(nums)):
            # 1. Push current element
            heapq.heappush(heap, (-nums[i], i))

            # 2. Remove elements outside window
            while heap[0][1] <= i - k:
                heapq.heappop(heap)

            # 3. Record result when window is ready
            if i >= k - 1:
                res.append(-heap[0][0])

        return res

        
```
