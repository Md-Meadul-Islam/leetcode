3719. Longest Balanced Subarray I

```python
from typing import List
class Solution:
    def longestBalanced(self, nums):
        n = len(nums)
        ans = 0

        for i in range(n):
            seen = set()
            even_count = 0
            odd_count = 0

            for j in range(i, n):
                if nums[j] not in seen:
                    seen.add(nums[j])

                    if nums[j] % 2 == 0:
                        even_count += 1
                    else:
                        odd_count += 1

                if even_count == odd_count:
                    ans = max(ans, j - i + 1)

        return ans
    
    def longestBalanced(self, nums: List[int]) -> int:
        n = len(nums)
        res = 0
        seen = set()
        for i in range(n):
            seen.clear()
            balance = 0
            if res > n - i:
                break
            for j in range(i, n):
                num = nums[j]
                if num not in seen: 
                    if num % 2:
                        balance -= 1
                    else:
                        balance += 1
                    seen.add(num)
                if balance == 0:
                    res = max(res, j - i + 1)
        
        return res

        
```

LEETCODE: HOW TO DETECT PATTERNS IN <60 SECONDS

## Step 1 — Identify the data structure

Array/string → sliding window, prefix sum, two pointers
Tree → DFS/BFS/recursion
Graph → BFS/DFS/Union-Find
Matrix → BFS/DFS
Linked list → two pointers

Always ask:
"What is the input structure?"

## Step 2 — Look for keywords

"subarray" → nested loop or sliding window
"distinct" → set or hashmap
"frequency/count" → hashmap
"range query" → prefix sum / segment tree
"shortest path" → BFS
"all combinations" → backtracking
"max/min subarray" → Kadane or prefix sum
"sorted" → binary search / two pointers

## Step 3 — Check constraints

n ≤ 1000      → O(n²) is OK
n ≤ 10^5      → O(n log n) or O(n)
n ≤ 10^6      → O(n) only

Constraints eliminate wrong approaches quickly.

## Step 4 — Ask the core question

Are we:

* enumerating subarrays?
* counting something?
* tracking uniqueness?
* optimizing a value?
* exploring paths?
* partitioning data?

Every problem reduces to one of these.

## Step 5 — Map to a known pattern

Common mapping:

subarray + count        → prefix sum
subarray + distinct     → set/hashmap
window + condition      → sliding window
range queries + update  → segment tree / Fenwick
choices/explore         → backtracking
levels/shortest path    → BFS
dependencies            → topological sort
merge intervals         → sorting
repeated subproblem     → DP

## Step 6 — Build smallest working solution

Start with brute force if allowed by constraints.
Then optimize only if necessary.

Most interview problems allow brute force first.

Key rule:
"If n ≤ 1500, try O(n²) first."
"If n ≤ 200, try O(n³)."

## Step 7 — Mental translation habit

Translate English → code concepts immediately.

Example:
"distinct even integers equals distinct odd integers"

becomes

even_set = set()
odd_set = set()

That translation skill is pattern recognition.

## Final insight

Understanding the description is NOT the goal.
Recognizing the pattern is the goal.

Strong problem solvers don't read faster —
they translate faster.
