```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        hash_map = {}
        max_len = 0
        left = 0
        for i in range(len(s)):
            if s[i] in hash_map:
                left = hash_map[s[i]] + 1
            max_len = max(max_len, i -left + 1)
            hash_map[s[i]] = i
        return max_len
```

```python
obj = Solution()
obj.lengthOfLongestSubstring("abcabcbb")
```
