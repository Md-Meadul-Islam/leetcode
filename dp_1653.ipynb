1653. Minimum Deletions to Make String Balanced

```python
class Solution:
    def minimumDeletions_dp(self, s: str) -> int:
        n = len(s)
        dp = [0] * n
        b_count = 0

        for i in range(n):
            if s[i] == 'b':
                b_count += 1
                dp[i] = dp[i-1] if i > 0 else 0
            else:
                if i == 0:
                    dp[i] = 0
                else:
                    dp[i] = min(dp[i-1] + 1, b_count)

        return dp[-1]
    
    def minimumDeletions_prefix_suffix(self, s: str) -> int:
        n = len(s)

        prefix_b = [0] * (n + 1)
        suffix_a = [0] * (n + 1)

        for i in range(n):
            prefix_b[i+1] = prefix_b[i] + (s[i] == 'b')

        for i in range(n-1, -1, -1):
            suffix_a[i] = suffix_a[i+1] + (s[i] == 'a')
        print(prefix_b, suffix_a)

        ans = float('inf')

        for i in range(n+1):
            ans = min(ans, prefix_b[i] + suffix_a[i])

        return ans
    def minimumDeletions_stack(self, s: str) -> int:
        stack = []
        deletions = 0

        for ch in s:
            if ch == 'b':
                stack.append('b')
            else:  # 'a'
                if stack:
                    stack.pop()
                    deletions += 1

        return deletions
    
    def minimumDeletions(self, s: str) -> int:
        b_count = 0
        deletions = 0

        for ch in s:
            if ch == 'b':
                b_count += 1
            else:  # ch == 'a'
                deletions = min(deletions + 1, b_count)

        return deletions

        
```
