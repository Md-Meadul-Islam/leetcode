1861. Rotating the Box


```python
from typing import List
class Solution:
    def rotateTheBox2(self, boxGrid: List[List[str]]) -> List[List[str]]:
        rows = len(boxGrid)
        cols = len(boxGrid[0])

        for r in range(rows):
            empty = cols - 1 # assume last column is empty
            for c in range(cols-1, -1, -1):
                if boxGrid[r][c] == "*":
                    empty = c - 1 # find obstacle, reset the column to current column
                elif boxGrid[r][c] == "#":
                    boxGrid[r][c] = "."
                    boxGrid[r][empty] = "#"
                    empty -= 1 #set pointer to prev column

        rotatedGrid = [["." for _ in range(rows)] for _ in range(cols)]

        for r in range(rows):
            for c in range(cols):
                rotatedGrid[c][rows-r-1] = boxGrid[r][c]
        return rotatedGrid
    
    def rotateTheBox2(self, box: List[List[str]]) -> List[List[str]]:
        r, c=len(box), len(box[0])
        rotate=[['.']*r for _ in range(c)]
        for i, row in enumerate(box):
            bottom=c-1
            for j in range(c-1, -1, -1):
                if row[j]=='#':
                    rotate[bottom][r-1-i]='#'
                    bottom-=1
                elif row[j]=='*':
                    rotate[j][r-1-i]='*'
                    bottom=j-1
        return rotate

            
```

```python
obj = Solution()
boxGrid =[
    ["#","#","*",".","*","."],
    ["#","#","#","*",".","."],
    ["#","#","#",".","#","."]]
obj.rotateTheBox(boxGrid)
```
