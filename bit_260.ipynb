260. Single Number III


```python
from typing import List
class Solution:
    def singleNumber(self, nums: List[int]) -> List[int]:
        xor = 0
        for num in nums:
            xor ^= num
        diff = xor & (-xor)
        a = 0
        b = 0
        for num in nums:
            if num & diff:
                a ^= num
            else:
                b ^= num
        return [a, b]
    
    def singleNumber2(self, nums: List[int]) -> List[int]:
        res=[]
        count=set()
        for i in nums:
            if i in count:
                res.remove(i)
            else:
                res.append(i)
                count.add(i)
        return res
    
    def singleNumber3(self, nums):
        count = {}

        for num in nums:
            count[num] = count.get(num, 0) + 1

        return [k for k, v in count.items() if v == 1]
```

```python
obj = Solution()
obj.singleNumber([1,2,1,3,2,5])
```
