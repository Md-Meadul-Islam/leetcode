137. Single Number II


```python
from typing import List
class Solution:
    def singleNumber2(self, nums: List[int]) -> int:
        result = 0

        for i in range(32):
            count = 0
            for num in nums:
                if (num>>i) & 1:
                    count += 1

            if count % 3 != 0:
                if i == 31:
                    result = result - (1<<31)
                else:
                    result = result | (1 << i)
        return result | 0
    def singleNumber(self, nums):
        ones = 0
        twos = 0
        for num in nums:
            ones = (ones ^ num) & ~twos
            twos = (twos ^ num) & ~ones

        return ones        
```

```python
obj = Solution()
obj.singleNumber2([-2,-2,1,1,4,1,4,4,-4,-2])
```
