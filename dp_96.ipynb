96. Unique Binary Search Tree

```python
import math
class Solution:
    def numTrees(self, n: int) -> int:
        dp = [0] * (n + 1)
        
        dp[0] = 1
        dp[1] = 1
        
        for nodes in range(2, n + 1):
            for root in range(1, nodes + 1):
                left = root - 1
                right = nodes - root
                dp[nodes] += dp[left] * dp[right]
        
        return dp[n]
    
    def numTrees_math(self, n: int) -> int:
        return math.comb(2*n, n) // (n+1)
```

```python
obj = Solution()
obj.numTrees_math(3)
```
