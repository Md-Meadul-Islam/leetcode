2833. Furthest Point From Origin


```python
class Solution:
    def furthestDistanceFromOrigin_brute_force(self, moves: str) -> int:
        l = 0
        r = 0
        for c in moves:
            if c == "L" or c == "_":
                l += 1
            else:
                l -= 1
        for c in moves:
            if c == 'R' or c == '_':
                r += 1
            else:
                r -= 1

        return max(abs(l), abs(r))
    def furthestDistanceFromOrigin(self, moves: str) -> int:
        left = 0
        right= 0
        blank = 0
        for ch in moves:
            if ch == "L":
                left += 1
            elif ch == "R":
                right += 1
            else:
                blank += 1
        print(left, right, blank)
        total = (left - right) 
        return abs(total) + blank

        
```
