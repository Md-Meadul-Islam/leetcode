2463. Minimum Total Distance Traveled


```python
from typing import List
from functools import lru_cache
class Solution:
    def minimumTotalDistance_DP(self, robot: List[int], factory: List[List[int]]) -> int:
        robot.sort()
        factory.sort()
        robots, factories = len(robot), len(factory)

        @lru_cache(None)
        def dp(r, f):
            if r == robots:
                return 0
            if f == factories:
                return float("inf")
            res = dp(r, f + 1)
            pos, limit = factory[f]
            cost = 0
            for k in range(1, limit + 1):
                if r + k > robots:
                    break
                cost += abs(robot[r+k-1] - pos)
                res = min(res, cost + dp(r+k, f+1))
            return res
        return dp(0, 0)
    
    def minimumTotalDistance_Tabulation(self, robot: List[int], factory: List[List[int]]) -> int:
        robot.sort()
        factory.sort()
        robots, factories = len(robot), len(factory)
        INF = float("inf")
        dp = [[INF] * (factories + 1) for _ in range(robots + 1)]

        for j in range(factories+1):
            dp[robots][j] = 0

        print(dp)
        for f in range(factories-1, -1, -1):
            pos, limit = factory[f]
            for r in range(robots, -1, -1):
                dp[r][f] = dp[r][f+1]
                cost = 0
                for k in range(1, limit+1):
                    if r + k >robots:
                        break
                    cost += abs(robot[r+k-1] - pos)
                    dp[r][f] = min(dp[r][f], cost + dp[r+k][f+1])
        print(dp)
        return dp[0][0]


        
```
