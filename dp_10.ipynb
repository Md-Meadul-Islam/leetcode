10. Regular Expression Matching


```python
class Solution:
    def isMatchDot(self, s, p):
    
        def match(i, j):
            
            if j == len(p):
                return i == len(s)
            
            if i < len(s) and (p[j] == s[i] or p[j] == '.'):
                return match(i + 1, j + 1)
            
            return False
        return match(0, 0)
    
    def isMatchStar(self, s, p):
        
        def match(i, j):
            
            if j == len(p):
                return i == len(s)
            
            first_match = (
                i < len(s) and 
                (p[j] == s[i] or p[j] == '.')
            )
            
            # If next character is '*'
            if j + 1 < len(p) and p[j + 1] == '*':
                
                # Option 1: skip x*
                # Option 2: use one match if possible
                return (
                    match(i, j + 2) or 
                    (first_match and match(i + 1, j))
                )
            
            else:
                return first_match and match(i + 1, j + 1)
        
        return match(0, 0)

    def isMatch_recursion(s: str, p: str) -> bool:
        
        def dfs(i, j):
            # if pattern finished
            if j == len(p):
                return i == len(s)
            
            # check if current character matches
            first_match = (
                i < len(s) and
                (p[j] == s[i] or p[j] == '.')
            )
            
            # if next character is '*'
            if j + 1 < len(p) and p[j + 1] == '*':
                
                # case 1: skip x*
                # case 2: use one occurrence (if match)
                return (
                    dfs(i, j + 2) or
                    (first_match and dfs(i + 1, j))
                )
            
            else:
                return first_match and dfs(i + 1, j + 1)
        
        return dfs(0, 0)
    
    def isMatch_memoization(s: str, p: str) -> bool:
        
        memo = {}
        
        def dfs(i, j):
            
            if (i, j) in memo:
                return memo[(i, j)]
            
            if j == len(p):
                return i == len(s)
            
            first_match = (
                i < len(s) and
                (p[j] == s[i] or p[j] == '.')
            )
            
            if j + 1 < len(p) and p[j + 1] == '*':
                
                result = (
                    dfs(i, j + 2) or
                    (first_match and dfs(i + 1, j))
                )
            
            else:
                result = first_match and dfs(i + 1, j + 1)
            
            memo[(i, j)] = result
            return result
        
        return dfs(0, 0)

    def isMatch_dp(self, s: str, p: str) -> bool:
        m, n = len(s), len(p)
        
        dp = [[False] * (n + 1) for _ in range(m + 1)]
        dp[0][0] = True
        
        # Handle empty string vs pattern like a*, a*b*, etc.
        for j in range(2, n + 1):
            if p[j - 1] == '*':
                dp[0][j] = dp[0][j - 2]
        
        for i in range(1, m + 1):
            for j in range(1, n + 1):
                
                # Case 1: normal char or '.'
                if p[j - 1] == s[i - 1] or p[j - 1] == '.':
                    dp[i][j] = dp[i - 1][j - 1]
                
                # Case 2: '*'
                elif p[j - 1] == '*':
                    
                    # zero occurrence
                    dp[i][j] = dp[i][j - 2]
                    
                    # one or more occurrence
                    if p[j - 2] == s[i - 1] or p[j - 2] == '.':
                        dp[i][j] |= dp[i - 1][j]
        
        return dp[m][n]

```
