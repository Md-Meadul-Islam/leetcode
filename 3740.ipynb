```python
from typing import List
from collections import defaultdict
class Solution:
    def minimumDistance_brute_force(self, nums: List[int]) -> int:
        hash_map = {}
        for i, num in enumerate(nums):
            arr = hash_map.get(num, [])
            arr.append(i)
            hash_map[num] = arr
        distances = []
        for key in hash_map:
            arr = hash_map[key]
            arr_len = len(arr)
            if arr_len >= 3:
                for i in range(arr_len - 2):
                    min_dis = min(arr[i], arr[i+1], arr[i+2])
                    max_dis = max(arr[i], arr[i+1], arr[i+2])

                    distances.append(2 * (max_dis - min_dis))
        return min(distances) if len(distances) else -1
    def minimumDistance(self, nums):
        dictionary = defaultdict(list)
        res = float('inf')

        for i, num in enumerate(nums):
            dictionary[num].append(i)
            
            if len(dictionary[num] >= 3):
                a, b, c = dictionary[num]
                res = min(res, 2*(c- a))
                dictionary[num].pop(0)
        return res if res != float("inf") else -1


    

        
```
